<b>Format string</b><br>
:< 5 - left align with 5 character <br>
:> 5 - right align with 5 character <br>
:^ 5 - center align with 5 character <br>
{-5:=8} - 8 space between - and 5 <br>
{-5:+} - always show the sign (+ or -) <br>
{-5:} - only show the sign for - <br>

```python
#alignment
name = "Bunga"
print(f"{f'Hello {name}':>20}")  #right align with 10 character
print(f"{f'Hello {name}':<20}")  #left align with 10 character
print(f"{f'Hello {name}':^20}")  #center align with 10 character
```

```python
#fill character
print(f"{'Hi':*<7}")  #fill with * (Hi + * = 7)
print(f"{'Hi':->5}")  #fill with - (Hi + - = 5)
print(f"{'Hi':~^6}")  #fill with ~ (Hi + ~ = 6)
```

```python
#sign handling
num = -12
num2 = 34

print("Always show sign")
print(f"{num:+}")
print(f"{num2:+}")

print("\nOnly show negative")
print(f"{num:}")
print(f"{num2:}")

print("\nPadding between sign and digit")
print(f"{num:=8}")
print(f"{num2:=8}")
```
