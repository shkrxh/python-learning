```python
#input
#by default string input
name = input("Enter your name: ")  
print(name)

age = input("Enter your age: ")
print(age, type(age))

age2 = int(input("Enter your age: "))  #integer
print(age2, type(age2))

price = float(input("Enter the price: "))  #float
print(price, type(price))
```

```python
#multiple input
a = input("First name: ")
b = input("Last name: ")
print(f"Hello {a} {b}!")

c = int(input("Enter value a: "))
d = int(input("Enter value b: "))
print(f"{c=} and {d=}")
```

.split() <br>
- use to enter multiple input using one line input <br>
.split() - input1 space input2 <br>
.split(",") - input1 comma input2 <br>
.split("-") - input1 dash input2 <br>
.split("|") - input1 pipe input2 <br>
.split(":") - input1 colon input2 <br>

```python
e, f = map(int, input("Enter two numbers: ").split())  
#use .split() to enter multiple input using one line input
#enter first number then clikc space before entering the second number
print(f"{e=} and {f=}")
```

```python
name, age =  input("Enter you name and age: ").split()
age = int(age)
print(f"{name} is {age} years old")
```

```python
data = input("Enter fruits that you know: ").split(", ")
print(data)
print(type(data))
```

```python
data = input("Enter a few numbers: ").split(":") #list of string input
number = list(map(int, data))  #convert to string to int and put in a list
print(data)
print(number)
```
